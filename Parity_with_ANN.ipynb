```python
import numpy as np
import pandas as pd

# Function to calculate even parity bit
def even_parity_bit(data):
    return np.sum(data) % 2

# Generate dataset
data = []
labels = []

# Generate all possible 8-bit combinations
for i in range(256):
    binary = list(map(int, format(i, '08b')))
    parity = even_parity_bit(binary)
    data.append(binary)
    labels.append(parity)

# Convert to pandas DataFrame
df = pd.DataFrame(data, columns=[f'bit_{i}' for i in range(8)])
df['parity'] = labels

# Display the first few rows of the DataFrame
df.head()


```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X = df.drop('parity', axis=1)
y = df['parity']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)



# Define the neural network model
model = Sequential([
    Dense(2000, input_dim=8, activation='relu'),  # Input layer with 2000 neurons and hidden layer with 16 neurons
    Dense(1000, activation='relu'),     # Hidden layer with 1000 neurons
    Dense(1, activation='softmax')             # Output layer with 1 neuron (parity bit)
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(X_train, y_train, epochs=100, batch_size=16, validation_split=0.2, verbose=1)



```

```python
# Function to predict and verify parity
def predict_parity(data):
    binary = np.array(list(map(int, format(data, '08b')))).reshape(1, -1)
    predicted_parity = (model.predict(binary) > 0.5).astype("int32")
    actual_parity = even_parity_bit(binary[0])
    print(f"Input Data: {format(data, '08b')}")
    print(f"Predicted Parity: {predicted_parity[0][0]}")
    print(f"Actual Parity: {actual_parity}")
    print("Verification:", "Pass" if predicted_parity[0][0] == actual_parity else "Fail")

# Example: Predict and verify parity for a new 8-bit input
predict_parity(100)  # Example input: 100 (binary: 01100100)
```
